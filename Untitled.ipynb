```python
import yfinance as yf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
import plotly.colors as pc
from datetime import datetime, timedelta
from dateutil.relativedelta import relativedelta




################################### CALCUL DES RENDEMENTS LP ET DCA ###################################

def calcul_rendement(duree_invest = 1 , somme_investie = 10000, mois_dca = 6, ticker = "^GSPC"):
    
#=============================== On prépare les variables ===============================  
    # Somme à investir par mois
    somme_par_mois = somme_investie / mois_dca

    # Définir les dates de début et de fin
    date_debut = datetime.now() - relativedelta(months=((duree_invest * 12) + 1)) # +1 pour s'assurer d'avoir un mois entier
    date_fin = datetime.now()

    # Télécharger les données financières pour la période
    data_financiere = yf.download(ticker, start=date_debut, end=date_fin, interval="1mo", auto_adjust=True)

    
#=============================== On clean le df ===============================
     # Remplace les cases vides par '0'
    if data_financiere.empty:
        return 0, 0
    # Remplace les cases NaN par '0'
    data_financiere = data_financiere.fillna(0)


    # Rendements mensuels en % (avec colonne 'Close' du df 'data_financiere')
    rendements_mois = data_financiere['Close'].pct_change().dropna()

    # Ajout colonne rendemenbt mensuel au df data_financiere
    data_financiere['Rendement du mois'] = rendements_mois

    
#=============================== On enlève le multi-index du df ===============================
    # Liste des colonnes à garder (niveau 'Price')
    colonnes_a_garder = ['Close', 'High', 'Low', 'Open', 'Volume', 'Rendement du mois']
    
    # Sélectionne uniquement ces colonnes dans une copie
    df_rendement = data_financiere.loc[:, data_financiere.columns.get_level_values(0).isin(colonnes_a_garder)].copy().dropna()
    
    # Aplatir les colonnes : transformer MultiIndex en noms simples
    df_rendement.columns = df_rendement.columns.get_level_values(0)

#=============================== On calcul le DCA et le LumpSum ===============================
    colonne_rendement = df_rendement['Rendement du mois']

    # Calcul du rendement DCA
    rendements_dca = []
    portefeuille_dca = 0
    # Investissement mensuel pendant la période DCA
    for i in range(min(mois_dca, len(colonne_rendement))):
        portefeuille_dca += somme_par_mois * (1 + colonne_rendement.iloc[i])
        rendements_dca.append(portefeuille_dca)
   
    # Croissance après la période DCA
    for i in range(mois_dca, len(colonne_rendement)):
        portefeuille_dca *= (1 + colonne_rendement.iloc[i])
        rendements_dca.append(portefeuille_dca)

    # Calcul du rendement LP
    rendements_lumpsum = []
    portefeuille_lumpsum = somme_investie
    for i in range(len(colonne_rendement)):
        portefeuille_lumpsum *= (1 + colonne_rendement.iloc[i])
        rendements_lumpsum.append(portefeuille_lumpsum)
    
    # On aligne la taille avec df_rendement, car pct_change() enlève le premier mois
    data_financiere = data_financiere.iloc[1:]  # on enlève le premier mois (NaN dans pct_change)
    data_financiere['Rendement LP'] = rendements_lumpsum
    data_financiere['Rendement DCA x mois'] = rendements_dca

    #display(data_financiere)

    return data_financiere

################################### AUTOMATISATION CALCULS RENDEMENTS pour graphique barre ###################################

def generer_df_rendements(duree_invest, somme_investie, mois_dca, ticker):

    #(duree_invest = 1 , somme_investie = 10000, mois_dca = 6, ticker = "^GSPC"):
    resultats = {'Année': []}

    # Initialiser dynamiquement les colonnes DCA
    for dca_mois in mois_dca:
        resultats[f'DCA ({dca_mois} mois)'] = []

    resultats['LumpSum'] = []

    # Boucle sur les durées d’investissement
    for duree in duree_invest:
        resultats['Année'].append(duree)

        # Calculer le rendement pour chaque durée DCA
        for dca_mois in mois_dca:
            dca_value= calcul_rendement(
                duree_invest=duree,
                somme_investie=somme_investie,
                mois_dca=dca_mois,
                ticker=ticker
            )
            resultats[f'DCA ({dca_mois} mois)'].append(dca_value)

        # Calculer le rendement LumpSum (mois_dca arbitraire ici)
            lump_value = calcul_rendement(
            duree_invest=duree,
            somme_investie=somme_investie,
            mois_dca=mois_dca,
            ticker=ticker
        )
        resultats['LumpSum'].append(lump_value)
        df_resultat1 = pd.DataFrame(resultats)
    # Retourner le DataFrame final
    return df_resultat1



################################### AUTOMATISATION CALCULS RENDEMENTS ###################################

def calcul_multiple_rendements(duree_invest, somme_investie, mois_dca, ticker):
    resultats = []
    
    for mois in mois_dca:
        for duree in duree_invest:
            df = calcul_rendement(
                duree_invest=duree,
                somme_investie=somme_investie,
                mois_dca=mois,
                ticker=ticker
            )
            df = df.copy()
            df["Durée"] = f"{duree} ans"
            df["Mois DCA"] = f"{mois} mois"
            resultats.append(df)

    df_resultat = pd.concat(resultats)

    # Supression des colonnes inutiles
    colonnes_a_supprimer = ['Open', 'High', 'Low', 'Volume']
    df_resultat = df_resultat.reset_index()
    df_resultat = df_resultat.drop(columns=colonnes_a_supprimer)
    df_resultat = df_resultat.round(4)
    df_resultat.to_csv("rendement_comparatif.csv", index=False)

    return df_resultat


################################### GRAPH BARRE ###################################
#calcul_multiple_rendements(duree_invest = [1, 5] , somme_investie = 10000, mois_dca = [6, 24], ticker = "^GSPC")
generer_df_rendements(duree_invest = [1, 5] , somme_investie = 10000, mois_dca = [6, 24], ticker = "^GSPC")
```

```python

############################################# Calcul des rendements DCA et Lump Sum #############################################

def calcul_rendement_mois(temps_investissement = 1 , somme_investie = 10000, mois_dca = 6, ticker = "^GSPC"):
    
#=============================== On prépare les variables ===============================  
    # Somme à investir par mois
    somme_par_mois = somme_investie / mois_dca

    # Définir les dates de début et de fin
    date_debut = datetime.now() - relativedelta(months=((temps_investissement * 12) + 1)) # +1 pour s'assurer d'avoir un mois entier
    date_fin = datetime.now()

    # Télécharger les données financières pour la période
    data_financiere = yf.download(ticker, start=date_debut, end=date_fin, interval="1mo", auto_adjust=True)

    
#=============================== On clean le df ===============================
     # Remplace les cases vides par '0'
    if data_financiere.empty:
        return 0, 0
    # Remplace les cases NaN par '0'
    data_financiere = data_financiere.fillna(0)


    # Rendements mensuels en % (avec colonne 'Close' du df 'data_financiere')
    rendements_mois = data_financiere['Close'].pct_change().dropna()

    # Ajout colonne rendemenbt mensuel au df data_financiere
    data_financiere['Rendement du mois'] = rendements_mois

    
#=============================== On enlève le multi-index du df ===============================
    # Liste des colonnes à garder (niveau 'Price')
    colonnes_a_garder = ['Close', 'Rendement du mois']
    
    # Sélectionne uniquement ces colonnes dans une copie
    df_rendement = data_financiere.loc[:, data_financiere.columns.get_level_values(0).isin(colonnes_a_garder)].copy().dropna()
    
    # Aplatir les colonnes : transformer MultiIndex en noms simples
    df_rendement.columns = df_rendement.columns.get_level_values(0)

    return df_rendement
    
    
#round(portefeuille_dca, 1), round(portefeuille_lumpsum, 1)


    
#=============================== On calcul le DCA et le LumpSum ===============================
def calcul_rdt_mois_chaque_duree(mois_dca, durations, somme_investie, ticker):
    # Calcul du rendement DCA
    portefeuille_dca = 0
    df_rendement = calcul_rendement_mois(temps_investissement = durations , somme_investie = somme_investie, mois_dca = mois_dca, ticker = ticker)
    somme_par_mois = somme_investie / mois_dca
    colonne_rendement = df_rendement['Rendement du mois']
    
    # Investissement mensuel pendant la période DCA
    for i in range(min(mois_dca, len(colonne_rendement))):
        portefeuille_dca += somme_par_mois * (1 + colonne_rendement.iloc[i])
    
    # Croissance après la période DCA
    for i in range(mois_dca, len(colonne_rendement)):
        portefeuille_dca *= (1 + colonne_rendement.iloc[i])

    # Calcul du rendement LumpSum
    portefeuille_lumpsum = somme_investie
    for i in range(len(colonne_rendement)):
        portefeuille_lumpsum *= (1 + colonne_rendement.iloc[i])

    round(portefeuille_dca, 1), round(portefeuille_lumpsum, 1) # On arrondit les valeurs

    df_rendement['Rendement x DCA'] = portefeuille_dca
    df_rendement['Rendement LS'] = portefeuille_lumpsum
    
    return df_rendement
    
    



def calcul_rdt_duree(dca_list, durations, somme_investie, ticker):
    resultats = {
    'Année': [],
    'Date de début': []
    }

    # Initialiser dynamiquement les colonnes DCA
    for dca_mois in dca_list:
        resultats[f'DCA ({dca_mois} mois)'] = []

    # Initialiser la colonnes LP
    resultats['LumpSum'] = []

    # Boucle sur les durées d’investissement
    for duree in durations:
        resultats['Année'].append(duree)


        # Calcul de la date de début pour cette durée
        date_debut = datetime.now() - relativedelta(months=(duree * 12))
        resultats['Date de début'].append(date_debut.strftime("%Y-%m-%d"))
        
        # Calculer le rendement pour chaque durée DCA
        for dca_mois in dca_list:
            dca_value, _ = calcul_rendement_mois(
                temps_investissement=duree,
                somme_investie=somme_investie,
                mois_dca=dca_mois,
                ticker=ticker
            )
            resultats[f'DCA ({dca_mois} mois)'].append(dca_value)

        # Calculer le rendement LumpSum (mois_dca arbitraire ici)
        _, lump_value = calcul_rendement_mois(
            temps_investissement=duree,
            somme_investie=somme_investie,
            mois_dca=dca_mois,
            ticker=ticker
        )
        resultats['LumpSum'].append(lump_value)
        
    df_resultats = pd.DataFrame(resultats)
    # Retourner le DataFrame final
    return df_resultats


calcul_rendement_mois(temps_investissement = 15 , somme_investie = 10000, mois_dca = 6, ticker = "^GSPC")
#df_resultats = calcul_rdt_mois_chaque_duree(mois_dca = 22, durations= 11, somme_investie= 10000, ticker= "^GSPC")
df = calcul_rdt_duree(dca_list = range(5, 11), durations = [3, 22], somme_investie = 100000, ticker="^GSPC")
df
```

```python
def graphe_barre(df_resultats):
    import plotly.graph_objects as go

    fig = go.Figure()

    # Couleurs personnalisées
    couleurs = [
        '#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b',
        '#e377c2', '#7f7f7f', '#bcbd22', '#17becf'
    ]

    # Détecter les colonnes DCA dynamiquement
    dca_colonnes = [col for col in df_resultats.columns if col.startswith("DCA")]

    # Ajout des barres DCA
    for i, col in enumerate(dca_colonnes):
        fig.add_trace(go.Bar(
            x=df_resultats["Année"],
            y=df_resultats[col],
            name=col,
            marker_color=couleurs[i % len(couleurs)]
        ))

    # Ajout des barres Lump Sum si présente
    if "LumpSum" in df_resultats.columns:
        fig.add_trace(go.Bar(
            x=df_resultats["Année"],
            y=df_resultats["LumpSum"],
            name="Lump Sum",
            marker_color="#000000"
        ))

    # Configuration du graphique
    fig.update_layout(
        barmode='group',  # ou 'stack' si tu veux empiler les barres
        title="Comparaison des rendements DCA vs Lump Sum sur différentes périodes",
        xaxis_title="Durée de l'investissement (années)",
        yaxis_title="Valeur finale (€)",
        legend_title="Méthode d'investissement",
        template="plotly_white",
        font=dict(family="Arial", size=14),
        height=800,
        width=1200,
        xaxis=dict(
            tickmode='array',
            tickvals=list(df_resultats["Année"]),
            tickangle=45,
            autorange='reversed'
        )
    )

    
    return fig.show()


graphe_barre(df_resultats)
```

```python
def graphe_line(t, somme_investie=10000):
    
    if t.empty:
        print("Erreur : Le DataFrame est vide")
        return
    
    df = t.copy().reset_index()
    df['Date'] = pd.to_datetime(df['Date'])
    df['Rendement LP'] = pd.to_numeric(df['Rendement LP'], errors='coerce')
    df['Rendement DCA x mois'] = pd.to_numeric(df['Rendement DCA x mois'], errors='coerce')
    
    fig = go.Figure()
    
    couleurs_dca = pc.qualitative.Bold
    couleurs_lump = pc.qualitative.Dark24
    durees = sorted(df['Durée'].unique())

    trace_info = []  # Pour stocker (durée, nom_trace)

    
    # Traces DCA
    for i, (duree, mois) in enumerate(df.groupby(['Durée', 'Mois DCA'])):
        if mois['Mois DCA'].iloc[0] != 1:  # Exclure LumpSum
            nom_trace = f"DCA {mois['Mois DCA'].iloc[0]} - {mois['Durée'].iloc[0]} ans"
            
            fig.add_trace(go.Scatter(
                x=mois['Date'],
                y=mois['Rendement DCA x mois'],
                mode='lines',
                name=nom_trace,
                line=dict(width=1.5, dash='dash', color=couleurs_dca[i % len(couleurs_dca)]),
                hovertemplate="Date: %{x}<br>Valeur: %{y:,.0f}€<extra></extra>"
            ))
            trace_info.append(mois['Durée'].iloc[0])  # Durée

    # Traces LumpSum
    for j, duree in enumerate(df['Durée'].unique()):
        for mois_dca in df['Mois DCA'].unique():
            df_filtered = df[(df['Durée'] == duree) & (df['Mois DCA'] == mois_dca)]
        nom_trace = f"LumpSum - {duree} ans"
        
        fig.add_trace(go.Scatter(
            x=df_filtered['Date'],
            y=df_filtered['Rendement LP'],
            mode='lines',
            name=nom_trace,
            line=dict(width=2, color=couleurs_lump[j % len(couleurs_lump)]),
            hovertemplate="Date: %{x}<br>Valeur: %{y:,.0f}€<extra></extra>"
        ))
        trace_info.append(duree)  # Durée

    
    # Création des boutons avec filtrage par durée (DCA + LumpSum)
    boutons_menu = []

    for duree in durees:
        visible_traces = [d == duree for d in trace_info]
        boutons_menu.append(dict(
            label=f"{duree} ans",
            method="update",
            args=[
                {"visible": visible_traces},
                {"title": f"Performance DCA vs LumpSum - {duree} ans"}
            ]
        ))

    # Bouton "Tout voir"
    boutons_menu.insert(0, dict(
        label="Tout voir",
        method="update",
        args=[
            {"visible": [True] * len(trace_info)},
            {"title": f"Performance DCA vs LumpSum (Investissement: {somme_investie:,.0f}€)"}
        ]
    ))

    # Layout
    fig.update_layout(
        title=dict(
            text=f"Performance DCA vs LumpSum (Investissement: {somme_investie:,.0f}€)",
            x=0.5,
            y=0.95,  # Monte le titre
            xanchor='center',
            yanchor="top",
            font=dict(size=20)
        ),
        xaxis=dict(
            title="Date",
            tickangle=-45,
            tickformat="%Y",
            dtick="M12",  # Un tick tous les 12 mois
            showgrid=True,
            gridcolor="LightGrey"
        ),
        yaxis_title="Valeur du portefeuille (€)",
        legend_title="Stratégie",
        template="plotly_white",
        hovermode="x unified",
        height=700,
        width=1200,
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=-0.5,
            xanchor="center",
            x=0.5
        ),
        updatemenus=[
            dict(
                type="buttons",  # Boutons côte à côte
                direction="right",
                buttons=boutons_menu,
                showactive=True,
                x=0.5,
                xanchor="center",
                y=1.1,
                yanchor="top"
            )
        ]
    )
    fig.show()

    return fig

graphe_line(df_resultats, somme_investie=10000)
```

```python
import yfinance as yf
import pandas as pd
from datetime import datetime
from dateutil.relativedelta import relativedelta

def rdt_par_mois_par_duree(durees, somme_investie, mois_dca_list, ticker):
    resultats = []

    for mois_dca in mois_dca_list:
        for duree in durees:
            #=============================== Variables ===============================  
            somme_par_mois = somme_investie / mois_dca
            date_debut = datetime.now() - relativedelta(months=((duree * 12) + 1))
            date_fin = datetime.now()

            data_financiere = yf.download(ticker, start=date_debut, end=date_fin, interval="1mo", auto_adjust=True)
            if data_financiere.empty:
                continue  # On saute les cas vides

            data_financiere = data_financiere.fillna(0)
            rendements_mois = data_financiere['Close'].pct_change().dropna()
            data_financiere['Rendement du mois'] = rendements_mois

            colonnes_a_garder = ['Close', 'High', 'Low', 'Open', 'Volume', 'Rendement du mois']
            df_rendement = data_financiere.loc[:, data_financiere.columns.get_level_values(0).isin(colonnes_a_garder)].copy().dropna()
            df_rendement.columns = df_rendement.columns.get_level_values(0)

            colonne_rendement = df_rendement['Rendement du mois']

            #========================= Calcul DCA =========================
            rendements_dca = []
            portefeuille_dca = 0
            for i in range(min(mois_dca, len(colonne_rendement))):
                portefeuille_dca += somme_par_mois * (1 + colonne_rendement.iloc[i])
                rendements_dca.append(portefeuille_dca)
            for i in range(mois_dca, len(colonne_rendement)):
                portefeuille_dca *= (1 + colonne_rendement.iloc[i])
                rendements_dca.append(portefeuille_dca)

            #========================= Calcul Lump Sum =========================
            rendements_lumpsum = []
            portefeuille_lumpsum = somme_investie
            for r in colonne_rendement:
                portefeuille_lumpsum *= (1 + r)
                rendements_lumpsum.append(portefeuille_lumpsum)

            #========================= Résultats =========================
            df_final = df_rendement.iloc[1:].copy()  # enleve le premier mois (NaN)
            df_final['Rendement DCA'] = rendements_dca[1:]
            df_final['Rendement LS'] = rendements_lumpsum[1:]
            df_final['Durée'] = f"{duree} ans"
            df_final['Mois DCA'] = f"{mois_dca} mois"

            resultats.append(df_final)

    #========================= Agrégation =========================
    df_resultat = pd.concat(resultats).reset_index()
    colonnes_a_supprimer = ['Open', 'High', 'Low', 'Volume']
    df_resultat = df_resultat.drop(columns=colonnes_a_supprimer)
    df_resultat = df_resultat.round(4)
    df_resultat.to_csv("rendement_comparatif.csv", index=False)

    return df_resultat

df = rdt_par_mois_par_duree(durees = [25, 20, 15, 10,5], somme_investie  = 10000, mois_dca_list = [6, 12, 24, 48], ticker = "^GSPC")
```

```python
def graphe_line(df, somme_investie=10000):
    
    fig = go.Figure()
    
    couleurs_dca = pc.qualitative.Bold
    couleurs_lump = pc.qualitative.Dark24
    durees = sorted(df['Durée'].unique())

    trace_info = []  # Pour stocker (durée, nom_trace)

    
    # Traces DCA
    for i, (duree, mois) in enumerate(df.groupby(['Durée', 'Mois DCA'])):
        if mois['Mois DCA'].iloc[0] != 1:  # Exclure LumpSum
            nom_trace = f"DCA {mois['Mois DCA'].iloc[0]} - {mois['Durée'].iloc[0]} ans"
            
            fig.add_trace(go.Scatter(
                x=mois['Date'],
                y=mois['Rendement DCA'],
                mode='lines',
                name=nom_trace,
                line=dict(width=1.5, dash='dash', color=couleurs_dca[i % len(couleurs_dca)]),
                hovertemplate="Date: %{x}<br>Valeur: %{y:,.0f}€<extra></extra>"
            ))
            trace_info.append(mois['Durée'].iloc[0])  # Durée

    # Traces LumpSum
    for j, duree in enumerate(df['Durée'].unique()):
        for mois_dca in df['Mois DCA'].unique():
            df_filtered = df[(df['Durée'] == duree) & (df['Mois DCA'] == mois_dca)]
        nom_trace = f"LumpSum - {duree} ans"
        
        fig.add_trace(go.Scatter(
            x=df_filtered['Date'],
            y=df_filtered['Rendement LS'],
            mode='lines',
            name=nom_trace,
            line=dict(width=2, color=couleurs_lump[j % len(couleurs_lump)]),
            hovertemplate="Date: %{x}<br>Valeur: %{y:,.0f}€<extra></extra>"
        ))
        trace_info.append(duree)  # Durée

    
    # Création des boutons avec filtrage par durée (DCA + LumpSum)
    boutons_menu = []

    for duree in durees:
        visible_traces = [d == duree for d in trace_info]
        boutons_menu.append(dict(
            label=f"{duree} ans",
            method="update",
            args=[
                {"visible": visible_traces},
                {"title": f"Performance DCA vs LumpSum - {duree} ans"}
            ]
        ))

    # Bouton "Tout voir"
    boutons_menu.insert(0, dict(
        label="Tout voir",
        method="update",
        args=[
            {"visible": [True] * len(trace_info)},
            {"title": f"Performance DCA vs LumpSum (Investissement: {somme_investie:,.0f}€)"}
        ]
    ))

    # Layout
    fig.update_layout(
        title=dict(
            text=f"Performance DCA vs LumpSum (Investissement: {somme_investie:,.0f}€)",
            x=0.5,
            y=0.95,  # Monte le titre
            xanchor='center',
            yanchor="top",
            font=dict(size=20)
        ),
        xaxis=dict(
            title="Date",
            tickangle=-45,
            tickformat="%Y",
            dtick="M12",  # Un tick tous les 12 mois
            showgrid=True,
            gridcolor="LightGrey"
        ),
        yaxis_title="Valeur du portefeuille (€)",
        legend_title="Stratégie",
        template="plotly_white",
        hovermode="x unified",
        height=700,
        width=1200,
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=-0.5,
            xanchor="center",
            x=0.5
        ),
        updatemenus=[
            dict(
                type="buttons",  # Boutons côte à côte
                direction="right",
                buttons=boutons_menu,
                showactive=True,
                x=0.5,
                xanchor="center",
                y=1.1,
                yanchor="top"
            )
        ]
    )
    

    return fig.show()

graphe_line(df, somme_investie=10000)
```
