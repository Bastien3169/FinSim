```python
from bs4 import BeautifulSoup
import requests
import time
import pandas as pd
import numpy as np
import yfinance as yf
import html5lib
import os
import glob
```

```python
# Récupération les données historiques pour chaque entreprise (je passe par google colab car fonctionne pas ici ????)

# Charger le fichier et prendre la liste de tous les tikers yahoo des entreprises pour voir les infos hist sur yfinance
df = pd.read_csv("/Users/bastoch/Desktop/datas_csv/data_indices/indices_complets.csv", encoding="utf-8") #Mettre chemin du fichier
#display(df)
print(df.columns)
stocks_tickers_yahoo = df["Ticker_Yahoo_Finance"].to_list()

print(stocks_tickers_yahoo)

print(len(stocks_tickers_yahoo))
```

```python
# Liste pour stocker les DataFrames
dfs = []

for i in stocks_tickers_yahoo:
    try:
        # Récupération des données historiques
        hist = yf.Ticker(i).history(period="max", interval="1wk")

        # Récupérer le shortName via .info
        short_name = yf.Ticker(i).info.get('shortName', 'Non disponible')

        # Ajouter les colonnes 'Ticker' et 'ShortName'
        hist['Ticker'] = i
        hist['ShortName'] = short_name

        # Ajouter le DataFrame historique dans la liste
        dfs.append(hist)

    except Exception as e:
        print(f"Erreur lors de la récupération des données pour {i}: {e}")

# Concaténation de tous les DataFrames
df_final_indices = pd.concat(dfs)

# Exportation en CSV
df_final_indices.to_csv("historique_indices.csv", index=True, encoding='utf-8')

# Affichage des résultats
print(df_final_indices.head())
```

```python
# Enfin on clean le fichier !

# Charger le fichier principal
df = pd.read_csv("historique_indices.csv", encoding="utf-8")

# Charger le fichier contenant les correspondances de tickers
df1 = pd.read_csv("indices_complets.csv", encoding="utf-8")

# Supprimer colonnes inutiles
df = df.drop(columns=["Open", "High", "Low", "Volume", "Dividends", "Stock Splits", "Capital Gains"], errors="ignore")

# Convertir la colonne "Date" en format datetime et reformater en "JJ-MM-AAAA"
df["Date"] = pd.to_datetime(df["Date"], errors="coerce", utc=True).dt.strftime("%d-%m-%Y")

# Renommer la colonne
df = df.rename(columns={"Ticker": "Ticker Yahoo Finance"})

# Fusionner les DataFrames pour ajouter la colonne "Ticker"
df = df.merge(df1[["Ticker Yahoo Finance", "Ticker"]], on="Ticker Yahoo Finance", how="left")

# Arrondir la colonne "Close"
df["Close"] = df["Close"].round(2)

# Réorganiser les colonnes dans l'ordre souhaité
df = df[["Date", "Close", "Ticker", "Ticker Yahoo Finance", "ShortName"]]

# Enregistrer le fichier modifié
df.to_csv("historique_indices_clean.csv", index=False, encoding="utf-8")

# Afficher un aperçu
display(df.head())
```
